When you vary the input current, the resting membrane potential varies according to the specific input. For example, when the current that is injected is 0 nA, the resting membrane potential over time solely decreases as the lack of input current and the existence of leakage in ions results in a gradual decrease in membrane potential as the membrane potential slowly is restored back to the resting potential at -70 mV. Finally, in the last case, where the injected current is 10 nA, the provided current is large enough to offset the leakage while bringing the potential to a value large enough to depolarize and reach the action potential, leading to the spikes (37 in total due to the constant current input).

In the beginning, when I did not remove calls to nsim.set_mpotential, there was no phase offset. However, once removing this function call, I observe a phase offset between the neurons when they excite one another. This makes sense because if the two neurons were to start at the same membrane potential and are injected the same current, they would spike at the same time and thus there would not be any offset. However, if they have different initial membrane potentials, then they would spike at different times. If the two neurons did not start at the same membrane potential, then, the firing of one will encourage the firing of the other, leading to an amplificiation in the difference in firing times. However, for the inhibitory connections, the neurons tend to synchronize their spikes in both cases where I intitially set the membrane potential and when I do not. This hold true because if one neuron spikes first, then it will slow down the spiking of the other, and vice versa (when one spikes it slows down the other, allowing the spiking neuron to catch up). Over time, this will eventually align the spiking, which will cause them to spike at the same time.

**Problem 2e.**

```python
# Initializing the matrix

t = np.empty((10, 10))

for i in range(10):
    for j in range(10):
        if j == i:
            t[i, j] = 1
        elif j == i - 1:
            t[i, j] = -0.2
        elif j == i + 1:
            t[i, j] = -0.2
        else:
            t[i, j] = 0

# Plot matrix
for row in t:
    print(row)
```

From the results, we can see that the transfomration of the input vector amplified the differences in the data. For example, we can see that the data is uniform in the first five elements (10) and uniform in the last five elements (20). In the output vector, we can see that the greatest difference between corresponding elements is from the 5th to 6th element in the vector, as it goes from 14. This phenomenon is a mach band, as the 10x10 transformation matrix we created emphasizes differences in the input data. In contrast, when there is no difference in the input data (when neighbors are the same value - excluding the rightmost and leftmost element), then the output is uniform. as seen in the 6s and 12s in the output array. This phenomenon can also be seen in the above graph, where the output intensity is uniform from elements 1-3 and 6-8 and where there is a huge difference in intensity between element 4 and element 5.

```python
# Initializing the matrix

t2 = np.empty((10, 10))

for i in range(10):
    for j in range(10):
        if j == i:
            t2[i, j] = 1
        elif j == i - 1:
            t2[i, j] = 1/3
        elif j == i + 1:
            t2[i, j] = 1/3
        else:
            t2[i, j] = 0

# Plot matrix
for row in t2:
    print(row)
```

```python
# Creating the input matrix 1
m1 = np.array([10, 10, 10, 10, 10, 20, 20, 20, 20, 20])

# Plotting the matrix
plt.plot(m1)
```

```python
# Creating the input matrix 2
# Creating the matrix
m2 = np.array([10, 8, 10, 8, 10, 8, 10, 8, 10, 8])

# Plotting the matrix
plt.plot(m2)
```

```python
# Transforming the two inputs with the positive network 
out1 = np.dot(t2, m1)
out2 = np.dot(t2, m2)

print(out1)
print(out2)
```

```python
# Without boundaries
out1_cleaned = out1[1:9]
out2_cleaned = out2[1:9]

print(out1_cleaned)
print(out2_cleaned)
```

Qualitatively, this network tries to minimize differences in the overall data. For example, for the first vector input, the data had 5 10s followed by 5 20s.
